### Step 3. Code

Write one cell per part. The first cell sets up the random numbers for part (d).

```python
rng = np.random.default_rng(1)
```

(a) Load the data with pd.read_csv(AIRQ, index_col=0) as aq. Count the missing values in each column.

```python
aq = pd.read_csv(AIRQ, index_col=0)
aq.shape
```

```python
aq.isna().sum()
```

(b) Mean imputation. Fill each missing Ozone with the mean Ozone. Save the result as oz_mean.

```python
mean = aq.Ozone.mean()
mean
```

```python
oz_mean = aq.copy()
```

```python
oz_mean['Ozone'] = oz_mean['Ozone'].fillna(mean)
```

```python
aq.isna().sum(), oz_mean.isna().sum()
```

(c) Regression imputation. Let `ok` hold the rows with an Ozone value. Fit a line with `b, a = np.polyfit(ok['Temp'], ok['Ozone'], 1)`. The predicted ozone for a day is `a + b * Temp`. Fill each missing Ozone with its prediction and save the result as oz_reg.

```python
oz_reg = aq.copy()
oz_reg.columns
```

```python
ok = oz_reg[oz_reg['Ozone'].notna()]
ok.shape
```

```python
b, a = np.polyfit(ok['Temp'], ok['Ozone'], 1)
```

```python
oz_reg['Ozone'] = oz_reg['Ozone'].fillna(a + b * oz_reg['Temp'])
```

```python
oz_reg.shape
```

```python
oz_reg.isna().sum()
```

```python
oz_reg.head()
```

(d) Stochastic regression imputation. Let sigma be the standard deviation of `ok['Ozone'] - (a + b * ok['Temp'])`. Add random noise `rng.normal(0, sigma, len(aq))` to the predictions, then fill the missing values with them. Save the result as oz_stoch.

```python
rng = np.random.default_rng(42)   # seed so the results are reproducible

ok = aq[aq['Ozone'].notna()]
b, a = np.polyfit(ok['Temp'], ok['Ozone'], 1)

resid = ok['Ozone'] - (a + b * ok['Temp'])
sigma = resid.std()

pred_noisy = a + b * aq['Temp'] + rng.normal(0, sigma, len(aq))

oz_stoch = aq.copy()
oz_stoch['Ozone'] = oz_stoch['Ozone'].fillna(pred_noisy)
```

```python
oz_stoch
```

```python
# your three Series from Step 3
for s in [aq['Ozone'], oz_mean['Ozone'], oz_reg['Ozone'], oz_stoch['Ozone']]:
    print(round(s.mean(), 1), round(s.std(), 1),
          round(s.corr(aq['Temp']), 3), round(s.min(), 1))
```
