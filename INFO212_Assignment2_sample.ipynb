Write the code that goes from URL to records.

Your script must:

- collect the features you listed in Assignment 1

````python
# import the libraries
import csv
import requests
from bs4 import BeautifulSoup
````

````python
# the url to the data source
data_url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"
````

````python
# Request the page and handle a failure.
try:
    response = requests.get(data_url, headers={"User-Agent": "PopulationClassProject/1.0"}, timeout=30)
    response.raise_for_status()
except requests.RequestException as error:
    raise SystemExit(f"Could not download the page: {error}")
````

````python
# Find the population table in the HTML.
soup = BeautifulSoup(response.text, "html.parser")
table = soup.find("table", class_="wikitable")
if table is None:
    raise SystemExit("The population table was not found.")
````

````python
# Remove footnote markers such as [1] and [a].
for footnote in table.find_all("sup"):
    footnote.decompose()
````

````python
# Extract each row data

records = []

for row in table.find_all("tr"):
    cells = row.find_all("td", recursive=False)
    if len(cells) < 5:
        continue  # Skip the header or incomplete rows.

    location = cells[0].get_text(" ", strip=True)
    if location == "World" or not location:
        continue  # Skip the global total and unnamed rows.

    # Convert population to an integer
    try:
        population = int(cells[1].get_text(strip=True).replace(",", ""))
    except ValueError:
        print(f"Skipping {location}: invalid population")
        continue

    # Extract world share
    try:
        world_share_pct = float(cells[2].get_text(strip=True).replace("%", ""))
    except ValueError:
        world_share_pct = ""

    # Extract reference date
    reference_date = cells[3].get_text(" ", strip=True)

    # Extract source type
    source_type = cells[4].get_text(" ", strip=True)

    records.append([
        location, population, world_share_pct, reference_date, source_type
    ])
````

````python
# inspect the records
print(records[:3])
````

````python
# Save the records as a plain text CSV file.
with open("world_population.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow([
        "location", "population", "world_share_pct",
        "reference_date", "source_type"
    ])
    writer.writerows(records)
````
