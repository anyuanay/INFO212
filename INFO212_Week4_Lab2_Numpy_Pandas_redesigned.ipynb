## Run the following cell to create three lists with random numbers

```python
import random
import pandas as pd

random.seed(21)

# ls1 contains 10 random decimal numbers in the range of (0.00, 150.00), each of which with 2-digit precision.
ls1 = [round(random.uniform(0, 150), 2) for _ in range(10)]

# ls2 contains 10 random integers in the range of (1, 500).
ls2 = [random.randint(1, 500) for _ in range(10)]

# ls3 contains 10 random decimal numbers in the range of (0.05, 0.90),
# each of which with 2-digit precision and the last digit rounded to the nearest 0 or 5.
ls3 = [round(random.uniform(0.05, 0.90) * 20) / 20 for _ in range(10)]
ls1, ls2, ls3
```
