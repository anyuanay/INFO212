```python
sns.kdeplot(df["points"], fill=True, label="density")
plt.hist(df["points"], bins=np.arange(79.5, 101.5, 1), density=True,
         alpha=0.4, label="histogram")
plt.legend()
plt.show()

print("lowest:", df["points"].min(), "highest:", df["points"].max())
```

### Step 4. Reflect

- The pie slices add up to 100 percent. Of what? What share of all wines the pie leaves out, and name the number the pie hides.

```python
top10 = df["country"].value_counts().head(10)
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
top10.plot.pie(ax=axes[0])
top10.plot.bar(ax=axes[1])
plt.tight_layout()
print(top10.sum(), "of", len(df), "reviews")
```
